```python
import numpy as np
from scipy.signal import butter, filtfilt
import matplotlib.pyplot as plt

# 3차원 경로 데이터를 생성하거나 불러옵니다 (예: x, y, z 좌표)
np.random.seed(42)
t = np.linspace(0, 10, 500)  # 시간 축
x = np.sin(t) + np.random.normal(0, 0.1, len(t))  # 노이즈가 섞인 신호
y = np.cos(t) + np.random.normal(0, 0.1, len(t))
z = np.sin(2 * t) + np.random.normal(0, 0.1, len(t))

# Butterworth 필터 설계
def lowpass_filter(data, cutoff, fs, order=4):
    nyquist = 0.5 * fs  # 나이퀴스트 주파수
    normal_cutoff = cutoff / nyquist
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    filtered_data = filtfilt(b, a, data)
    return filtered_data


# 필터 적용
cutoff = 2.0  # 컷오프 주파수 (Hz)
fs = 50.0  # 샘플링 주파수 (Hz)
x_filtered = lowpass_filter(x, cutoff, fs)
y_filtered = lowpass_filter(y, cutoff, fs)
z_filtered = lowpass_filter(z, cutoff, fs)

# 필터링 결과 시각화
plt.figure(figsize=(12, 8))
plt.subplot(3, 1, 1)
plt.plot(t, x, label='Original X')
plt.plot(t, x_filtered, label='Filtered X', linewidth=2)
plt.legend()

plt.subplot(3, 1, 2)
plt.plot(t, y, label='Original Y')
plt.plot(t, y_filtered, label='Filtered Y', linewidth=2)
plt.legend()

plt.subplot(3, 1, 3)
plt.plot(t, z, label='Original Z')
plt.plot(t, z_filtered, label='Filtered Z', linewidth=2)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np


def split_timestamps(timestamps, interval=3):
    # 시작과 끝 시간을 계산
    start_time = np.floor(timestamps.min())
    end_time = np.ceil(timestamps.max())

    # 타임스탬프와 인덱스를 정렬
    sorted_indices = np.argsort(timestamps)
    sorted_timestamps = timestamps[sorted_indices]

    # 각 타임스탬프에 대한 빈 인덱스 계산
    bins = np.arange(start_time, end_time + interval, interval)
    bin_indices = np.digitize(sorted_timestamps, bins) - 1

    # 빈 인덱스가 변경되는 지점 찾기
    bin_change_indices = np.where(np.diff(bin_indices) != 0)[0] + 1
    splits = np.split(sorted_indices, bin_change_indices)

    # 그룹 생성
    groups = []
    for group_indices in splits:
        if len(group_indices) > 0:
            groups.append((group_indices[0], group_indices[-1]))

    return groups
```

```python
def split_timestamps_by_interval_numpy(timestamps, interval=5):
    """
    Split a numpy array of timestamps into groups where each group contains
    timestamps that are at most `interval` seconds apart.

    Args:
        timestamps (numpy.ndarray): A 1D numpy array of timestamps (assumed to be sorted).
        interval (int): The maximum time difference (in seconds) to group.

    Returns:
        list of numpy arrays: Groups of timestamps split by the given interval.
    """
    if timestamps.size == 0:
        return []

    timestamps = np.sort(timestamps)  # Ensure the timestamps are sorted
    diff = np.diff(
        timestamps
    )  # Calculate differences between consecutive timestamps
    print(diff)
    split_indices = (
        np.where(diff > interval)[0] + 1
    )  # Indices where the difference exceeds the interval

    # Split the array into groups
    groups = np.split(timestamps, split_indices)
    return split_indices


# Example
timestamps = np.array([1, 2, 5, 7, 10, 12, 15, 20])
print(split_timestamps_by_interval_numpy(timestamps))
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 예제 포인트 클라우드 데이터 생성 (실제 데이터로 대체하세요)
np.random.seed(0)
pointcloud = np.random.rand(100, 3)

# X 좌표를 독립 변수로 사용하고 Y, Z를 종속 변수로 사용
X = pointcloud[:, 0].reshape(-1, 1)
Y = pointcloud[:, 1]
Z = pointcloud[:, 2]

# Y에 대한 선형 회귀 모델 생성 및 학습
model_Y = LinearRegression()
model_Y.fit(X, Y)

# Z에 대한 선형 회귀 모델 생성 및 학습
model_Z = LinearRegression()
model_Z.fit(X, Z)

# 예측값 계산
X_fit = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
Y_fit = model_Y.predict(X_fit)
Z_fit = model_Z.predict(X_fit)

# 결과 시각화
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# 원래의 포인트 클라우드 플롯
ax.scatter(
    pointcloud[:, 0],
    pointcloud[:, 1],
    pointcloud[:, 2],
    color='blue',
    label='Point Cloud',
)

# 피팅된 직선 플롯
ax.plot(X_fit.flatten(), Y_fit, Z_fit, color='red', label='Fitted Line')

# 그래프 설정
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.legend()
plt.show()
```
